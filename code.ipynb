Business recommendations

```python
# Which region should the business focus on to increase profit?
res = df.groupby("Region")[["Sales", "Profit"]].sum()
res["Persentage"] = (res["Profit"] / res["Sales"] *100).round(2)
res.loc[res["Profit"].idxmax()]
```

```python
# Which products or sub-categories need a discount strategy review?
```

```python
res = df.groupby("Sub-Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Avg_Discount=("Discount", "mean")
)

res.sort_values(["Total_Profit", "Avg_Discount"], ascending=[True, False]).round(2)
```

```python
res[(res["Avg_Discount"] > df["Discount"].mean()) & (res["Total_Profit"] < 0)]
```

```python
res = df.groupby("Product Name").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Avg_Discount=("Discount", "mean")
)

res[(res["Avg_Discount"] > df["Discount"].mean()) & (res["Total_Profit"] < 0)].sort_values("Total_Profit")
```

```python
#What sales trends can you identify over time?
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly_sales = df.groupby(
    df["Order Date"].dt.to_period("M")
)["Sales"].sum().sort_index()

monthly_sales
```

```python
monthly_sales.pct_change().mul(100).round(2)
```

```python
monthly_sales.idxmax(), monthly_sales.max()
```

```python
df.groupby(
    df["Order Date"].dt.year
)["Sales"].sum().sort_index()
```

```python
monthly_sales.plot(
    kind="line",
    marker="o",
    title="Monthly Sales Trend",
    xlabel="Month",
    ylabel="Total Sales",
    grid=True
)
```

```python
#1. Review discounts on loss-making products
#Identify products with high discounts and negative profit to determine where discount reductions or pricing changes may help.
df.groupby("Sub-Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Avg_Discount=("Discount", "mean")
).sort_values("Total_Profit").round(2)
```
